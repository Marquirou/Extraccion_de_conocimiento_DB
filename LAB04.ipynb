```python
#Limpieza 
#Nombre de columnas en minusculas 
df.columns = df.columns.str.strip().str.lower()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["cabin"], inplace=True)
```

```python
df.columns
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df.isnull().sum()
```

```python
df["fare"] = df["fare"].fillna(df["fare"].mode()[0])
```

```python
df.isnull().sum()
```

```python
#Validar Datos 
df["age"].min()
```

```python
#Graficamos el total de pasajeros por clase
plt.figure(figsize=(8,4)) #Tamaño
sns.countplot(df, x="pclass")
plt.title("Total de pasajeros por clase")
plt.xlabel("Clase")
plt.ylabel("Total")
plt.show()
```

```python
#Graficamos las clases por género
plt.figure(figsize=(8,4))
sns.countplot(df, x="sex", hue="pclass")
plt.title("Total de Género por Clases")
plt.xlabel("CLase")
plt.ylabel("Género")
plt.show()              
```

```python
#Mostramos la edad por género y clase
plt.figure(figsize=(6,4))
sns.violinplot(df,x="sex", y="age", hue="pclass")
plt.title("Rango de Edad por Género y Clase")
plt.xlabel("Género")
plt.ylabel("Edad")
plt.show() 
```

```python
#Conocimiento
#Los pasajeros de primera clase pagaban tárifas más caras
df.groupby("pclass")["fare"].mean()
```
